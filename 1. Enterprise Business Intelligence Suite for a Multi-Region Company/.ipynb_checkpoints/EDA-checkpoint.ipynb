# Exploratory Data Analysis - EDA

```python
# Import & load libraries
import os 
import pandas as pd
```

```python
# View all columns 
pd.set_option('display.max_columns', None)
```

```python
# Import & Load Raw Dataset
df = pd.read_csv("Raw Data/IBM HR Analytics Employee Attrition & Performance.csv")
df.head()
```

```python
# Dataset Size
display(df.shape)
```

```python
# Analyzing Columns
display(df.columns)
```

```python
# Statistical Summary
display(df.describe().T)
```

```python
# Checking Data Types
display(df.dtypes)
```

```python
# Analyzing Null Values
display(df.isna().sum())
```

```python
# Analyzing Duplicated Values
display(df.duplicated().sum())
```

```python
# Checking Yes/No Values
print("\nAttrition distribution:\n", df["Attrition"].value_counts())
```

```python
# Analyzing StandardHours Columns
df['StandardHours'].unique()
```

```python
# Renaming Columns as per Snake case
df.columns = (
    df.columns.str.strip()
    .str.replace(r"(?<!^)(?=[A-Z])", "_", regex=True)
    .str.lower()
)
```

```python
df.head()
```

```python
# Changing Data Type from string to int 
df["attrition_flag"] = df["attrition"].eq("Yes").astype("int8")
```

```python
df.head()
```

# Summary Table 

```python
# Grouping Ages
df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 25, 35, 45, 55, float("inf")],
    labels=["18-25", "26-35", "36-45", "46-55", "56+"],
)
```

```python
# Grouping tenure_group
df["tenure_group"] = pd.cut(
    df["years_at_company"],
    bins=[-1, 1, 5, 10, float("inf")],
    labels=["0-1 years", "2-5 years", "6-10 years", "11+ years"],
)
```

```python
# Calculate prior experience outside current company
df["experience_gap"] = df["total_working_years"] - df["years_at_company"]
```

```python
# Drop constant or unhelpful columns
df = df.drop(
    columns=["employeecount", "over18", "standardhours"], errors="ignore"
)
```

```python
# Output Directory
output_dir = (
    "Summary"
)
```

```python
os.makedirs(output_dir, exist_ok=True)
```

```python
# Export cleaned data
df.to_csv(
    os.path.join(output_dir, "summary.csv"), index=False
)
```

```python
# Aggregate metrics by department and role for high-level overview
summary_df = (
    df.groupby(["department", "job_role"])
    .agg(
        total_employees=("employee_number", "count"),
        total_exits=("attrition_flag", "sum"),
        avg_monthly_income=("monthly_income", "mean"),
        avg_tenure=("years_at_company", "mean"),
    )
    .reset_index()
)
```

```python
# Compute attrition rate percentage
summary_df["attrition_rate_pct"] = round(
    (summary_df["total_exits"] / summary_df["total_employees"]) * 100, 2
)
```

```python
# Export the summary data to csv
summary_df.to_csv(os.path.join(output_dir, "summary.csv"), index=False)
```

```python
# Print confirmation and preview the final summary table
print(
    f"Summary table created and saved successfully to: {output_dir}/summary.csv"
)
print(summary_df.head())
```
