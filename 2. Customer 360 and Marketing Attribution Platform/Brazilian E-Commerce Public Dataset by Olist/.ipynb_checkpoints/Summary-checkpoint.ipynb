```python
orders = pd.read_csv(
    "Raw Data/olist_orders_dataset.csv"
)

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(
        orders[col],
        errors="coerce"
    )

orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400

orders["delivery_variance_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / 86400
```

```python



# ---------------------------------------
# 1. Create useful business columns
# ---------------------------------------

orders["purchase_year"] = orders["order_purchase_timestamp"].dt.year
orders["purchase_month"] = orders["order_purchase_timestamp"].dt.to_period("M").astype("string")
orders["purchase_day"] = orders["order_purchase_timestamp"].dt.day_name()

# Time taken for order approval
orders["approval_hours"] = (
    orders["order_approved_at"] - orders["order_purchase_timestamp"]
).dt.total_seconds() / 3600

# Actual delivery status
orders["is_delivered"] = (
    orders["order_status"].eq("delivered")
)

# Delivery performance
orders["delivery_performance"] = "Not Delivered"

delivered_mask = (
    orders["order_delivered_customer_date"].notna()
    & orders["order_estimated_delivery_date"].notna()
)

orders.loc[delivered_mask, "delivery_performance"] = (
    orders.loc[delivered_mask, "delivery_variance_days"]
    .apply(lambda x: "Late" if x > 0 else ("On Time" if x == 0 else "Early"))
)

# ---------------------------------------
# 2. Select important columns
# ---------------------------------------

summary_columns = [
    "order_id",
    "customer_id",
    "order_status",
    "order_purchase_timestamp",
    "purchase_year",
    "purchase_month",
    "purchase_day",
    "order_approved_at",
    "approval_hours",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "delivery_days",
    "delivery_variance_days",
    "is_delivered",
    "delivery_performance"
]

order_summary = orders[summary_columns].copy()

# ---------------------------------------
# 3. Create dataset overview
# ---------------------------------------

overview_data = [
    ["Total Records", len(orders), "Total number of order records"],
    ["Unique Orders", orders["order_id"].nunique(), "Number of distinct orders"],
    ["Unique Customers", orders["customer_id"].nunique(), "Distinct customer IDs represented in orders"],
    ["Duplicate Order IDs", orders["order_id"].duplicated().sum(), "Repeated order IDs"],
    ["First Purchase Date", orders["order_purchase_timestamp"].min(), "Earliest purchase timestamp"],
    ["Last Purchase Date", orders["order_purchase_timestamp"].max(), "Latest purchase timestamp"],
    ["Delivered Orders", orders["order_status"].eq("delivered").sum(), "Orders with delivered status"],
    ["Cancelled Orders", orders["order_status"].eq("canceled").sum(), "Orders with canceled status"],
    ["Average Delivery Days", orders["delivery_days"].mean(), "Average purchase-to-delivery duration"],
    ["Median Delivery Days", orders["delivery_days"].median(), "Median purchase-to-delivery duration"],
    ["Average Delivery Variance", orders["delivery_variance_days"].mean(), "Positive means late; negative means early"],
    ["Missing Purchase Dates", orders["order_purchase_timestamp"].isna().sum(), "Records without purchase timestamps"],
    ["Missing Customer Delivery Dates", orders["order_delivered_customer_date"].isna().sum(), "Records without actual customer delivery dates"]
]

dataset_overview = pd.DataFrame(
    overview_data,
    columns=["Metric", "Value", "Description"]
)

# Add order status distribution
status_counts = (
    orders["order_status"]
    .value_counts(dropna=False)
    .rename_axis("Metric")
    .reset_index(name="Value")
)

status_counts["Description"] = "Number of orders with this status"
status_counts = status_counts[["Metric", "Value", "Description"]]

dataset_overview = pd.concat(
    [dataset_overview, status_counts],
    ignore_index=True
)

# ---------------------------------------
# 4. Create column dictionary
# ---------------------------------------

column_dictionary = pd.DataFrame([
    ["order_id", "Unique order identifier", "ID"],
    ["customer_id", "Customer identifier linked to the customer dataset", "ID"],
    ["order_status", "Current order status", "Category"],
    ["order_purchase_timestamp", "Date and time when the order was placed", "Datetime"],
    ["purchase_year", "Year in which the order was placed", "Date"],
    ["purchase_month", "Year and month of purchase", "Date"],
    ["purchase_day", "Day of the week when the order was placed", "Date"],
    ["order_approved_at", "Date and time when payment was approved", "Datetime"],
    ["approval_hours", "Hours between purchase and approval", "Numeric"],
    ["order_delivered_carrier_date", "Date when the order was handed to the carrier", "Datetime"],
    ["order_delivered_customer_date", "Actual date of customer delivery", "Datetime"],
    ["order_estimated_delivery_date", "Estimated customer delivery date", "Datetime"],
    ["delivery_days", "Actual days between purchase and customer delivery", "Numeric"],
    ["delivery_variance_days", "Actual delivery date minus estimated delivery date; positive means late", "Numeric"],
    ["is_delivered", "True when order status is delivered", "Boolean"],
    ["delivery_performance", "Early, On Time, Late, or Not Delivered", "Category"]
], columns=["Column Name", "Business Meaning", "Data Type"])

# ---------------------------------------
# 5. Create data quality report
# ---------------------------------------

data_quality = pd.DataFrame({
    "Column Name": orders.columns,
    "Data Type": orders.dtypes.astype(str).values,
    "Missing Values": orders.isna().sum().values,
    "Missing Percentage": (
        orders.isna().mean().values * 100
    ).round(2),
    "Unique Values": orders.nunique(dropna=True).values
})

# ---------------------------------------
# 6. Export all sheets to Excel
# ---------------------------------------

output_file = "Olist_Order_Summary_Report.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    order_summary.to_excel(
        writer, sheet_name="Order Summary", index=False
    )

    dataset_overview.to_excel(
        writer, sheet_name="Dataset Overview", index=False
    )

    column_dictionary.to_excel(
        writer, sheet_name="Column Dictionary", index=False
    )

    data_quality.to_excel(
        writer, sheet_name="Data Quality", index=False
    )

print("Summary report created successfully!")
print(f"File: {output_file}")
print(f"Total order records: {len(order_summary):,}")
```
